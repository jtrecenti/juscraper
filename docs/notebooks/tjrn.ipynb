# TJRN - Tribunal de Justica do Rio Grande do Norte

Jurisprudence scraper for TJRN, via Elasticsearch API.

| Feature | Available |
|---------|----------|
| cjsg    | Yes      |

```python
import juscraper as jus

tjrn = jus.scraper('tjrn')
```

## Basic search

```python
df = tjrn.cjsg('dano moral', paginas=range(1, 3))

print(df.shape)
df.head(3)
```

## Available columns

```python
brutos = tjrn.cjsg_download('dano moral', paginas=1)
df_parsed = tjrn.cjsg_parse(brutos)

print(df_parsed.shape)
df_parsed.head(3)
```

## Preview ementa

```python
print(df['ementa'].iloc[0][:300])
```

## Using filters

Available filters: `numero_processo`, `id_classe`, `id_orgao_julgador`, `id_relator`, `id_colegiado`, `data_julgamento_inicio`/`data_julgamento_fim`, `sistema`, `decisoes`, `jurisdicoes`, `grau`, `buscar_em`.

`sistema`, `decisoes`, `jurisdicoes` and `grau` only take the values of the portal form, or `""`/`None` for all; any other value raises `ValidationError` before the request:

- `sistema`: `"PJE"` or `"SAJ"` (case-sensitive).
- `decisoes`: `"Decisão"` (single-judge), `"Acórdão"` (collegiate) or `"Sentença"` (first instance).
- `jurisdicoes`: `"TJ"` (Tribunal de Justiça) or `"TR"` (Turmas Recursais); both only cover second-instance decisions.
- `grau`: `"1"` or `"2"`, as strings.

By default `pesquisa` searches the ementa, and only acórdãos have one: with `decisoes="Decisão"`, `decisoes="Sentença"` or `grau="1"`, search the term with `buscar_em="inteiro_teor"` (see below).

```python
df_filtered = tjrn.cjsg(
    'direito',
    data_julgamento_inicio='2024-01-01',
    data_julgamento_fim='2024-06-30',
    paginas=1,
)

print(df_filtered.shape)
df_filtered.head(3)
```

## Searching the full text

`buscar_em` picks where to look for `pesquisa`: `"ementa"` (default) or `"inteiro_teor"`, the full text of the decision (the portal's "Inteiro teor" field), which acórdãos, single-judge decisions and sentenças all have.

```python
df_sentencas = tjrn.cjsg('dano moral', buscar_em='inteiro_teor', decisoes='Sentença', paginas=1)

print(df_sentencas.shape)
df_sentencas[['processo', 'classe', 'orgao_julgador', 'data_julgamento']].head(3)
```
