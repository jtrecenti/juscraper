# TJRO - Tribunal de Justica de Rondonia

Jurisprudence scraper for TJRO, via Elasticsearch API (JURIS).

| Feature | Available |
|---------|----------|
| cjsg    | Yes      |

```python
import juscraper as jus

tjro = jus.scraper('tjro')
```

## Basic search

```python
df = tjro.cjsg('dano moral', paginas=1)

print(df.shape)
df.head(3)
```

## Available columns

```python
brutos = tjro.cjsg_download('dano moral', paginas=1)
df_parsed = tjro.cjsg_parse(brutos)

print(df_parsed.shape)
df_parsed.head(3)
```

## Using filters

Available filters: `tipo` (list), `numero_processo`, `relator`, `orgao_julgador`, `orgao_julgador_colegiado`, `classe`, `instancia`, `termo_exato` (bool), and the date range `data_julgamento_inicio` / `data_julgamento_fim`.

The old names `nr_processo`, `magistrado`, `classe_judicial` and `data_inicio` / `data_fim` are still accepted as deprecated aliases and emit a `DeprecationWarning`.

```python
df_filtered = tjro.cjsg(
    'direito',
    data_julgamento_inicio='2024-01-01',
    data_julgamento_fim='2024-06-30',
    paginas=1,
)

print(df_filtered.shape)
df_filtered.head(3)
```

## Repeated rows

The portal sometimes indexes the same decision twice, as two documents with different ids
(`_id` and `id_processo_documento` in the raw response) and identical content. Both come back in the
search, next to each other, so `cjsg` can return rows that are equal in every column; when the pair
falls on a page boundary, the repeated row shows up on two consecutive pages. Measured on
2026-11-16: 4 such pairs in the first 400 results of `'dano moral'`, with no repeated `_id`.
The scraper returns what the portal returns. The column `id_documento` carries the `_id` of each
document, so the two rows of a pair differ only there: a repeated `id_documento` would be the same
document downloaded twice, while equal rows with different ids are the portal's own duplicates. If
you need one row per decision, drop them with
`df.drop_duplicates(subset=df.columns.drop('id_documento'))`.
