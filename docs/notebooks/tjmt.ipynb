## Basic search

```python
import juscraper as jus

tjmt = jus.scraper('tjmt')

dados_cjsg = tjmt.cjsg('dano moral', paginas=1)
print(dados_cjsg.shape)
dados_cjsg.head(3)
```

## Using filters

You can filter by publication date range (`data_publicacao_inicio`/`data_publicacao_fim`, end date inclusive), judge (`relator`), court chamber (`orgao_julgador`), case type (`tipo_processo`), and case class (`classe`). The API does not filter by judgment date: passing `data_julgamento_*` raises `TypeError`. `relator`, `orgao_julgador` and `classe` take the exact name, not a numeric id: `classe="Apelação Cível"` and `classe="Apelação"` are different classes.

```python
dados_filtrados = tjmt.cjsg(
    'direito',
    paginas=1,
    data_publicacao_inicio='2025-01-01',
    data_publicacao_fim='2025-06-30',
    tipo_processo='Criminal',
)
print(dados_filtrados.shape)
dados_filtrados.head(3)
```

## Querying different types

Besides `Acordao` (default), you can also query `DecisaoMonocratica`.

The `id` column is the API's `idDocumento`, which comes as `0` for older single-judge decisions (`origem == 'Proteus'`). To deduplicate, use `uuid`, which is distinct for every document.

```python
dados_decisao = tjmt.cjsg('direito', paginas=1, tipo_consulta='DecisaoMonocratica')
print(dados_decisao.shape)
dados_decisao.head(3)
```

## Download and parse separately

You can also download the raw data first and parse it later.

```python
brutos = tjmt.cjsg_download('direito', paginas=1)
print(f"Pages downloaded: {len(brutos)}")
print(f"Keys in response: {list(brutos[0].keys())}")

dados = tjmt.cjsg_parse(brutos)
import pandas as pd
df = pd.DataFrame(dados)
print(f"\nParsed shape: {df.shape}")
df.head(3)
```
