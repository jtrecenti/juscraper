# TJSC - Tribunal de Justica de Santa Catarina

Jurisprudence scraper for TJSC, via the eproc PHP system.

| Feature | Available |
|---------|----------|
| cjsg    | Yes      |

```python
import juscraper as jus

tjsc = jus.scraper('tjsc')
```

## Basic search

```python
df = tjsc.cjsg('dano moral', paginas=range(1, 3))

print(df.shape)
df.head(3)
```

## Available columns

```python
brutos = tjsc.cjsg_download('dano moral', paginas=1)
df_parsed = tjsc.cjsg_parse(brutos)

print(df_parsed.shape)
df_parsed.head(3)
```

## Preview ementa

```python
if 'ementa' in df.columns and len(df) > 0:
    print(df['ementa'].iloc[0][:300])
```

## Using filters

Available filters: `buscar_em` (`"ementa"` or `"inteiro_teor"`), `processo`, `data_julgamento_inicio/fim`, `data_publicacao_inicio/fim`. Dates accept `YYYY-MM-DD` or `DD/MM/YYYY`.

```python
df_filtered = tjsc.cjsg(
    'direito',
    buscar_em='ementa',
    paginas=1,
)

print(df_filtered.shape)
df_filtered.head(3)
```

### Filter by judgment date

`data_julgamento_inicio` and `data_julgamento_fim` restrict the search to decisions judged in the interval (both ends inclusive).

```python
df_datas = tjsc.cjsg(
    'direito',
    paginas=1,
    data_julgamento_inicio='2025-03-10',
    data_julgamento_fim='2025-03-13',
)

print(df_datas.shape)
print(df_datas['data_julgamento'].min(), df_datas['data_julgamento'].max())
df_datas[['processo', 'orgao_julgador', 'data_julgamento']].head(3)
```
