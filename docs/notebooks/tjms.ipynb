# TJMS - Tribunal de Justica de Mato Grosso do Sul

Scraper for the jurisprudence searches of the Tribunal de Justica de Mato Grosso do Sul, second-instance (CJSG) and first-instance (CJPG), and for the PDFs of the decisions they return.

| Feature | Available |
|---------|----------|
| cjsg    | Yes      |
| cjpg    | Yes      |
| baixar_acordao / baixar_acordaos | Yes |
| baixar_sentenca / baixar_sentencas | Yes |
| cpopg   | No       |
| cposg   | No       |

## Basic search

```python
import juscraper as jus

tjms = jus.scraper('tjms')
```

```python
df = tjms.cjsg('dano moral', paginas=range(1, 3))
print(df.shape)
df.head(3)
```

## Using filters

You can filter by judgment date, publication date, decision type, and other fields.

```python
df_filtered = tjms.cjsg(
    'direito',
    data_julgamento_inicio='01/01/2024',
    data_julgamento_fim='30/06/2024',
    paginas=1,
)
print(df_filtered.shape)
df_filtered.head(3)
```

## Download and parse separately

You can also download the raw HTML files and parse them in a separate step.

```python
import tempfile, os

tmp = tempfile.mkdtemp()
pasta = tjms.cjsg_download('direito', paginas=1, diretorio=tmp)
print('Downloaded to:', pasta)
print('Files:', os.listdir(pasta))
```

```python
df_parsed = tjms.cjsg_parse(pasta)
print(df_parsed.shape)
df_parsed.head(3)
```

## First-instance decisions (cjpg)

`cjpg()` searches the first-instance decisions (sentenças) of TJMS. Unlike `cjsg`, `pesquisa` may be empty, to search by filters only (e.g. `classe`, `vara`, `data_julgamento_inicio`). Each row brings the text of the decision in `decisao` and the codes `baixar_sentencas` uses to download its PDF.

```python
sentencas = tjms.cjpg('dano moral', paginas=1)
print(sentencas.shape)
sentencas.head(3)
```

## Sentence PDFs (baixar_sentencas)

`baixar_sentencas()` downloads the PDF of the sentence in each `cjpg` row, without login or captcha, and adds `caminho_sentenca` (`None` when the download failed) and `motivo_falha`. `baixar_sentenca()` downloads a single one.

```python
pdfs = tjms.baixar_sentencas(sentencas.head(3), diretorio=tmp)
pdfs[['id_processo', 'caminho_sentenca', 'motivo_falha']]
```
