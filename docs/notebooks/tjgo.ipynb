# TJGO — Court of Justice of Goiás

Scraper for the TJGO jurisprudence search hosted on Projudi
(`ConsultaJurisprudencia`). The page renders a Cloudflare Turnstile
widget. Since 12 November 2026 (Projudi 3.194.1) the backend validates
its token, which juscraper does not send: every search comes back empty,
without an error from the portal, and `cjsg` emits a `UserWarning`
explaining the block. The cells below keep the usual workflow and
handle the empty result.

```python
import juscraper as jus

tjgo = jus.scraper('tjgo')
```

## Basic search

Download two pages of results so we can verify pagination works.

```python
df = tjgo.cjsg('dano moral', paginas=range(1, 3))
print(df.shape)
df.head(3)
```

## Available columns

```python
df.columns.tolist()
```

## Text preview

```python
if df.empty:
    print('No results: the portal requires the Turnstile token.')
else:
    print(df['texto'].iloc[0][:300])
```

## Filtering by instance and area

The search form exposes dropdowns that we can map directly:

* ``id_instancia``: ``0`` all / ``15`` tribunal / ``151`` appellate panels (turmas recursais)
* ``id_area``: ``0`` all / ``1`` civil / ``2`` criminal

```python
df_civel = tjgo.cjsg('dano moral', paginas=1, id_instancia=15, id_area=1)
df_civel[['processo', 'serventia', 'tipo_ato']].head() if not df_civel.empty else df_civel
```

## Download and parse separately

`cjsg_download` returns the raw HTML responses, which is useful when
you want to keep the original payloads.

```python
raw = tjgo.cjsg_download('dano moral', paginas=1)
print(type(raw), len(raw), 'chars in page 1:', len(raw[0]))
df_raw = tjgo.cjsg_parse(raw)
df_raw.head(3)
```
